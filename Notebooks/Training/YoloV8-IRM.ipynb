## Imports

```python
from ultralytics import YOLO
import time
```

## We define where our Yaml config files are located

```python
data_cfg_path = "/workspace/YoloV8-Tumor-Detection/data.yaml"
model_cfg_path = "/workspace/YoloV8-Tumor-Detection/configs/hyp_tumor.yaml"
```

## Now we define YoloV8's variant we will use

```python
model = YOLO("yolov8m.pt")
```

## Now we launch the training with only 1 epochs to test  and mesure time

```python
import os
os.chdir("/workspace/YoloV8-Tumor-Detection")
start = time.time()
model.train(
    data="Data/Raw/data.yaml",
    cfg="configs/hyp_tumor.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    workers=8,
    device=0,
    patience=20,
)

elapsed = time.time() - start
print(f"Training time: {elapsed / 60:.1f} min ({elapsed / 3600:.2f} h)")
```

## We print evaluation metrics afterwards

```python
metrics = model.val(data="data.yaml")
print(metrics.box.map50, metrics.box.map)
```
