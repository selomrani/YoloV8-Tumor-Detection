## Imports

```python
from ultralytics import YOLO
import time
```

## We define where our Yaml config files are located

```python
data_cfg_path = "../../Data/Raw/data.yaml"
model_cfg_path = "../../configs/hyp_tumor.yaml"
```

## Now we define YoloV8's variant we will use

```python
model = YOLO("yolov8m.pt")
```

## Now we launch the training with only 1 epochs to test  and mesure time

```python
start = time.time()
model.train(
    data="Data/data.yaml",
    cfg="configs/hyp_tumor.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    workers=8,
    device=0,
    patience=20,
)

elapsed = time.time() - start
print(f"Training time: {elapsed / 60:.1f} min ({elapsed / 3600:.2f} h)")
```
